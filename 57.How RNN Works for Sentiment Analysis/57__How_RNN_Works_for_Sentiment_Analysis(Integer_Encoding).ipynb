### **Why RNN?**  
RNN (Recurrent Neural Network) is used for sequential data (e.g., text, speech, time series) because it maintains memory of previous inputs, making it suitable for patterns that depend on order.  

### **How RNN Works:**  
1. **Sequential Processing** – Takes one input at a time (e.g., one word in a sentence).  
2. **Hidden State** – Stores past information and updates with each new input.  
3. **Weight Sharing** – Same weights are used at every time step, reducing complexity.  
4. **Backpropagation Through Time (BPTT)** – Adjusts weights based on errors over multiple time steps.  

**Issue:** Standard RNNs struggle with long sequences due to vanishing gradients, which LSTMs and GRUs solve by improving memory handling.

### Text preparation for RNN:  

1. **Tokenization** – Split text into words or subwords.  
2. **Build Vocabulary** – Create a list of unique words.  
3. **Integer Encoding** – Map each word to a unique integer.  
4. **Word Embeddings (Optional)** – Convert integers into dense vectors (like Word2Vec, GloVe).  
5. **Padding** – Ensure all sequences have the same length by adding zeros (post-padding or pre-padding).  

This makes the text numerical and uniform in shape, allowing RNNs to process it.

```python
# Import necessary modules
import numpy as np
import tensorflow as tf #added import
from tensorflow.keras.preprocessing.text import Tokenizer #Changed Import
from tensorflow.keras.utils import pad_sequences #changed import
from tensorflow.keras.datasets import imdb #Changed Import
from tensorflow.keras import Sequential #Changed Import
from tensorflow.keras.layers import Dense, SimpleRNN, Embedding, Flatten #Changed Import

# input text :
docs = [
    'make america bankrupt again',
    'fake news! fake news!',
    'biggest hands in history, believe me',
    'i know words, i have the best words',
    'covfefe was a message from the future',
    'nobody respects women more than me',
    'i am, like, a really smart person',
    'windmills cause cancer, folks',
    'china china china!',
    'mexico will pay for the wall... eventually'
]
```

```python
# Create a Tokenizer instance with an out-of-vocabulary (OOV) token
tokenizer = Tokenizer(oov_token='<nothing>')
# The 'oov_token' argument assigns '<nothing>' as a placeholder for words not present in the training dataset.
```

```python
tokenizer.fit_on_texts(docs)  # Updates the tokenizer with the given text dataset

# This step tokenizes the `docs` list by assigning a unique integer to each word.
# The tokenizer builds a vocabulary based on the words in `docs` and prepares them for further text processing.
```

```python
tokenizer.word_index
```

```python
tokenizer.word_counts  # Returns a dictionary containing words as keys and their corresponding frequencies as values

# This provides a count of how many times each word appears in the `docs` dataset.
# Example output (depending on `docs` content):
# {'make': 1, 'america': 1, 'bankrupt': 1, 'again': 1, 'fake': 2, 'news': 2, ...}
```

```python
tokenizer.document_count  # Returns the total number of documents (texts) used to fit the tokenizer

# This gives the number of documents in the dataset `docs`, which is the total number of strings or text entries.
# In this case, the output would be 10 since there are 10 documents in the `docs` list.
```

```python
sequences = tokenizer.texts_to_sequences(docs)  # Converts the list of documents to sequences of integers

# This method transforms each document in `docs` into a sequence of integers where each integer represents a word's index in the tokenizer's vocabulary.
# For example, if 'make' is the first word in the vocabulary, it will be represented by '1', 'america' by '2', and so on.
# The output will be a list of lists, where each sublist corresponds to a document's tokenized sequence.

```

```python
from keras.utils import pad_sequences
```

```python
from keras.preprocessing.sequence import pad_sequences  # Import pad_sequences for sequence padding

sequences = pad_sequences(sequences, padding='post')
# Pads sequences to ensure they all have the same length.
# 'padding="post"' means padding is added at the end of each sequence.
# By default, the padding value is 0.

# Example:
# Before padding:
# [[1, 2, 3], [4, 5], [6, 7, 8, 9]]
#
# After padding:
# [[1, 2, 3, 0],
#  [4, 5, 0, 0],
#  [6, 7, 8, 9]]
```

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train, y_train), (X_test, y_test) = imdb.load_data()
# Loads the IMDB dataset, which consists of preprocessed movie reviews for sentiment analysis.

# X_train, X_test: Lists of movie reviews represented as sequences of integers (word indices).
# y_train, y_test: Binary sentiment labels (0 = negative, 1 = positive).

# The dataset contains 50,000 reviews:
# - 25,000 for training (X_train, y_train)
# - 25,000 for testing (X_test, y_test)
```

```python
X_train[0]  # Displays the first movie review in the training dataset as a sequence of integers.

# The output will be a list of integers, where each integer represents a word from the IMDB dataset's predefined vocabulary.
# Example output (actual numbers may vary):
# [1, 14, 22, 16, 43, 530, 973, 1622, 1385, 65, 458, 4468, 66, 3941, 4, 173, 36, 256, 6, 630, 7, 4, 112,  6, 214, ...]

# Each number corresponds to a word in the IMDB dataset's dictionary, where lower numbers represent more frequent words.
```

```python
len(X_train[0])
```

```python
X_train = pad_sequences(X_train, padding='post', maxlen=50)
X_test = pad_sequences(X_test, padding='post', maxlen=50)

# Pads all sequences in X_train and X_test to ensure they have a uniform length of 50.
# 'padding="post"' means padding is added at the end of each sequence.
# 'maxlen=50' truncates longer reviews to 50 words and pads shorter ones with 0s.

# Example:
# Before padding/truncation:
# [1, 14, 22, 16, 43, 530, 973, 1622, 1385, 65, 458, 4468, 66, ...] (length > 50)
#
# After padding/truncation:
# [1, 14, 22, 16, 43, 530, 973, 1622, 1385, 65, ..., 0, 0, 0]  (length = 50)
```

```python
model = Sequential()  # Initializes a sequential model

model.add(SimpleRNN(32, input_shape=(50,1), return_sequences=False))
# Adds a SimpleRNN layer with 32 units
# 'input_shape=(50,1)' indicates input sequences of length 50 with 1 feature per time step
# 'return_sequences=False' means the RNN only returns the last output (suitable for classification tasks)

model.add(Dense(1, activation='sigmoid'))
# Adds a Dense (fully connected) layer with 1 neuron and sigmoid activation
# Used for binary classification (output between 0 and 1)

model.summary()
# Displays a summary of the model architecture, including layer types, output shapes, and parameter counts.
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
# Compiles the model for binary classification
# 'binary_crossentropy' is the loss function used for binary classification tasks
# 'adam' is an adaptive optimization algorithm
# 'accuracy' is used as the evaluation metric during training

model.fit(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
# Trains the model using the training data (X_train, y_train)
# 'epochs=5' means the model will go through the entire dataset 5 times
# 'validation_data=(X_test, y_test)' evaluates the model's performance on the test set after each epoch
```
