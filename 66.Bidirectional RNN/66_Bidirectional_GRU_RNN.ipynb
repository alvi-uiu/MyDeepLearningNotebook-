### Bidirectional RNN (BiRNN) :

#### **What is Bidirectional RNN?**  
A **Bidirectional RNN** processes a sequence **both forward and backward**, using two separate hidden states. This allows the model to capture **past (left context) and future (right context) dependencies** in the data.

#### **How it Works?**  
- Two RNN layers (e.g., GRU, LSTM) run **in parallel**:  
  - **Forward RNN** processes input from start to end.  
  - **Backward RNN** processes input from end to start.  
- Their outputs are **combined (concatenated or summed)** at each time step.  

#### **When to Use?**  
✅ **Text & Speech Processing** (e.g., Sentiment Analysis, Named Entity Recognition, Speech Recognition).  
✅ **Sequence Labeling** (e.g., POS Tagging, Machine Translation).  
✅ **Time-Series Forecasting** (if future context is available).  
🚫 **Not for real-time applications** (since it requires full sequence input).  

BiRNN improves performance by capturing **context from both past and future words**, making it powerful for NLP tasks. 🚀

```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Bidirectional, GRU, Dense

# Load the IMDB dataset
max_features = 10000  # Number of unique words to consider (most frequent words)
maxlen = 100  # Max length of each review (pad/truncate to this length)

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)

# Pad sequences to ensure uniform input size
x_train = pad_sequences(x_train, maxlen=maxlen)
x_test = pad_sequences(x_test, maxlen=maxlen)

# Define the Bidirectional GRU model
model = Sequential([
    Embedding(max_features, 32, input_length=maxlen),
    # Converts word indices into dense 32-dimensional word vectors.

    Bidirectional(GRU(5, return_sequences=True)),
    # First Bidirectional GRU layer with 5 units in each direction (total 10 outputs per time step).

    Bidirectional(GRU(5)),
    # Second Bidirectional GRU layer with 5 units in each direction (outputs a single 10-dimensional vector).

    Dense(1, activation='sigmoid')
    # Single neuron with sigmoid activation for binary classification (positive or negative sentiment).
])

# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model
model.fit(x_train, y_train, epochs=5, batch_size=32, validation_data=(x_test, y_test))

# Evaluate the model
loss, accuracy = model.evaluate(x_test, y_test)
print(f"Test Accuracy: {accuracy:.4f}")
```


### Explanation:
1. **Data Preparation**  
   - Loads the IMDB dataset (pre-tokenized).
   - Limits vocabulary size to **10,000** most frequent words.
   - Pads or truncates reviews to **100 words** for uniform input.

2. **Model Architecture**  
   - **Embedding Layer**: Converts word indices into 32-dimensional dense vectors.  
   - **Bidirectional GRU Layers**:  
     - The first layer **returns sequences** to allow another RNN layer to process the full sequence.  
     - The second layer **outputs a final representation** for classification.  
   - **Dense Output Layer**: Uses `sigmoid` activation for **binary sentiment classification**.

3. **Compilation & Training**  
   - Uses `adam` optimizer and `binary_crossentropy` loss.  
   - Trains for **5 epochs** with batch size **32**.  
   - Evaluates on the **test set**.

This model is designed for **sentiment analysis** on IMDB reviews using **Bidirectional RNN (GRU)**, capturing **both past and future context** in reviews. 🚀
