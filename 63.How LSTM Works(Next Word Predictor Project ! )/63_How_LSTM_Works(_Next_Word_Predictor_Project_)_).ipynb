LSTM (Long Short-Term Memory) networks are a type of recurrent neural network (RNN) designed to overcome the limitations of standard RNNs—especially the vanishing gradient problem—when learning long-term dependencies in sequence data. Here are the important points:

- **Memory Cells:**  
  LSTMs introduce a memory cell that maintains information over long periods. This cell acts as a “conveyor belt” that runs through the network with only minor linear interactions, helping preserve the gradient during training.

- **Gating Mechanisms:**  
  The key innovation in LSTMs is the use of gates, which regulate the flow of information into and out of the cell:
  - **Input Gate:** Controls how much new information from the current input is added to the cell state.
  - **Forget Gate:** Decides what information from the previous cell state should be discarded.
  - **Output Gate:** Determines which part of the cell state is output as the hidden state for the current time step.
  Each gate typically uses a sigmoid function, producing values between 0 (completely block) and 1 (completely allow).

- **Handling Long-Term Dependencies:**  
  By combining these gates with the cell state, LSTMs can effectively capture long-term dependencies in data (like language or time-series), since they can decide which information to keep, update, or remove over many time steps.

- **Advantages Over Traditional RNNs:**  
  The structure of LSTMs helps mitigate the vanishing gradient problem, enabling the network to learn from much longer sequences compared to traditional RNNs. This makes them highly effective for tasks such as language modeling, translation, speech recognition, and other sequence prediction tasks.

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([faqs])
# Updates the tokenizer's vocabulary based on the input text stored in 'faqs'.
# This converts the words in 'faqs' into a word index, assigning a unique integer to each word.
# If 'faqs' is a single string instead of a list of strings, it will be treated as a single text entry.

# Example (conceptual):
# Input:
# faqs = "how to reset my password"
# The tokenizer creates a word index:
# {'how': 1, 'to': 2, 'reset': 3, 'my': 4, 'password': 5}
```

```python
for sentence in faqs.split('\n'):
  print(tokenizer.texts_to_sequences([sentence])[0])
```

```python
input_sequences = []
# Initializes an empty list to store generated sequences

for sentence in faqs.split('\n'):
    # Splits the text (`faqs`) into individual sentences based on newline characters

    tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]
    # Converts the sentence into a sequence of integer tokens using the tokenizer
    # The output is a list of word indices representing the sentence

    for i in range(1, len(tokenized_sentence)):
        # Iterates through the tokenized sentence to create partial sequences

        input_sequences.append(tokenized_sentence[:i+1])
        # Appends progressively longer subsequences of the sentence
        # Each sequence starts from the first word and extends up to the i-th word

# Example (conceptual):
# Suppose `sentence = "hello world good morning"`
# Tokenized sentence: [5, 10, 15, 20]
# Generated `input_sequences`:
# [ [5, 10],
#   [5, 10, 15],
#   [5, 10, 15, 20] ]
```

```python
input_sequences
```

```python
max_len = max([len(x) for x in input_sequences])
max_len
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
# Imports the `pad_sequences` function from TensorFlow's Keras module

padded_input_sequences = pad_sequences(input_sequences, maxlen=max_len, padding='pre')
# Pads the sequences in `input_sequences` to ensure they all have the same length (`max_len`)
# 'padding="pre"' means that padding (0s) will be added at the beginning of each sequence
# This ensures that all sequences have a uniform length

# Example (conceptual):
# Before padding (assuming max_len=5):
# [[5, 10], [5, 10, 15], [5, 10, 15, 20]]
#
# After padding:
# [[ 0,  0,  0,  5, 10],
#  [ 0,  0,  5, 10, 15],
#  [ 0,  5, 10, 15, 20]]
```

```python
padded_input_sequences
```

```python
X = padded_input_sequences[:, :-1]
# Extracts all columns except the last one from `padded_input_sequences`
# This represents the input sequences used for training

y = padded_input_sequences[:, -1]
# Extracts only the last column from `padded_input_sequences`
# This represents the target (label) for each input sequence

# Example (conceptual):
# Suppose `padded_input_sequences` is:
# [[ 0,  0,  5, 10, 15],
#  [ 0,  5, 10, 15, 20],
#  [ 5, 10, 15, 20, 25]]

# After extraction:
# X (input):
# [[ 0,  0,  5, 10],
#  [ 0,  5, 10, 15],
#  [ 5, 10, 15, 20]]

# y (target):
# [15, 20, 25]
```

```python
from tensorflow.keras.utils import to_categorical
# Imports the `to_categorical` function, which converts integer labels into one-hot encoded vectors

y = to_categorical(y, num_classes=276)
# Converts the target labels (`y`) into one-hot encoded format with 276 possible classes
# This is useful for multi-class classification problems where each output corresponds to a word index

# Example (conceptual):
# Suppose `y` (before conversion) is:
# [15, 20, 25]

# After `to_categorical(y, num_classes=30)` (assuming 30 classes for visualization):
# [[ 0,  0,  0, ..., 0, 1, 0, ..., 0],   # 1 at index 15
#  [ 0,  0,  0, ..., 0, 1, 0, ..., 0],   # 1 at index 20
#  [ 0,  0,  0, ..., 0, 1, 0, ..., 0]]   # 1 at index 25

# Each row is a one-hot encoded representation of the corresponding integer label.
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
import numpy as np

# Create a Sequential model
model = Sequential()

# Embedding layer with input_shape specified
model.add(Embedding(276, 100, input_shape=(40,)))
# Embedding layer:
# - Input dimension: 276 (vocabulary size, i.e., number of unique words)
# - Output dimension: 100 (each word is mapped to a 100-dimensional vector)
# - input_shape: (40,) specifies that each input sequence has 40 elements
# This layer converts word indices into dense vectors, allowing the model to learn meaningful word representations.

model.add(LSTM(150, return_sequences=True))
# First LSTM layer with 150 units.
# LSTM (Long Short-Term Memory) is a type of recurrent neural network (RNN) designed to capture long-term dependencies in sequences.
# return_sequences=True ensures this layer outputs the full sequence for the next LSTM layer.

model.add(LSTM(150))
# Second LSTM layer with 150 units.
# Stacking multiple LSTM layers helps the model learn more complex patterns in sequential data.

model.add(Dense(276, activation='softmax'))
# Fully connected output layer:
# - 276 neurons (one for each word in the vocabulary)
# - Activation function: `softmax` (used for multi-class classification, as it outputs probability distributions)

# Compile the model
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

# Print model summary
model.summary()

# To test the model with dummy data (uncomment to test):
# x_dummy = np.random.randint(0, 276, size=(32, 40))  # 32 samples of length 40
# y_dummy = np.random.randint(0, 2, size=(32, 276))   # One-hot encoded targets
# model.fit(x_dummy, y_dummy, epochs=1, batch_size=32)
```

```python
model.fit(X,y,epochs=100)
```

```python
import time
# Imports the time module to control the delay between predictions.

text = "One Piece is a legendary"
# Initial input text that will be used for generating the sequence.

for i in range(20):
    # Loop to generate the next 10 words iteratively, extending the sequence.

    token_text = tokenizer.texts_to_sequences([text])[0]
    # Tokenizes the current `text` into a sequence of integer tokens using the tokenizer.

    padded_token_text = pad_sequences([token_text], maxlen=56, padding='pre')
    # Pads the tokenized sequence to a fixed length of 56 (with padding added at the beginning).
    # This ensures that all input sequences are of the same length, as expected by the model.

    pos = np.argmax(model.predict(padded_token_text))
    # Predicts the next word by passing the padded input sequence to the model.
    # `np.argmax` is used to get the index of the class with the highest predicted probability.

    for word, index in tokenizer.word_index.items():
        # Iterates through the word-to-index mapping in the tokenizer.

        if index == pos:
            # Checks if the predicted index matches the word index in the tokenizer.

            text = text + " " + word
            # Adds the predicted word to the current text.

            print(text)
            # Prints the updated text after adding the predicted word.

            time.sleep(2)
            # Pauses the execution for 2 seconds to create a delay between predictions.

# Example (conceptual):
# Initial text: "One Piece is a legendary"
# After the first iteration, it might become: "One Piece is a legendary anime"
# After the second iteration: "One Piece is a legendary anime series"
# And so on, generating the next words based on the model's predictions.
```
